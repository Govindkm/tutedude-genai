```python
def show_items(*items):
    print("items =", items)
    print("type  =", type(items))
    for item in items:
        print(f"item   = {item} (type={type(item)})")

show_items("pen", "book", "eraser")
show_items(10, 20)
```

```python
def Hello(name):
    """
    Greets the person who's name is given.
    Args:
        name (string): Name of the person whom to greet
    """
    print(f"Hello {name}")
```

```python
def is_palindrome(text:str):
    isStr = isinstance(text, str)
    if isStr != True:
        raise ValueError("Function excepts only string values")
    reverse = text[::-1]
    # print(reverse)
    if(reverse == text):
        return True
    
    return False
```

```python
print(is_palindrome("jahaj"))
print(is_palindrome("kukuk"))
print(is_palindrome("to_lower"))
print(is_palindrome(1))
```
