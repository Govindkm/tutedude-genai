# List in Python

> A list is an ordered and mutable data structure in Python.
> It can store multiple values, including mixed data types.

````python
# List Data Structure in Python
# A list is an ordered, mutable collection that can hold items of different types.

# 1. Creating Lists
empty_list = []
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True, None]
nested = [[1, 2], [3, 4], [5, 6]]

print("Empty list:", empty_list)
print("Numbers:", numbers)
print("Mixed types:", mixed)
print("Nested list:", nested)

# 2. Accessing Elements (Indexing & Slicing)
print("\n--- Accessing Elements ---")
print("First element:", numbers[0])       # Index starts at 0
print("Last element:", numbers[-1])       # Negative indexing
print("Slice [1:4]:", numbers[1:4])       # Slicing (start inclusive, end exclusive)

# 3. Modifying Lists (Mutable)
print("\n--- Modifying Lists ---")
numbers[0] = 10
print("After changing first element:", numbers)

# 4. Common List Methods
print("\n--- Common List Methods ---")
fruits = ["apple", "banana", "cherry"]

fruits.append("date")              # Add to end
print("After append:", fruits)

fruits.insert(1, "blueberry")      # Insert at index
print("After insert:", fruits)

fruits.remove("banana")            # Remove by value
print("After remove:", fruits)

popped = fruits.pop()              # Remove and return last element
print("Popped element:", popped)
print("After pop:", fruits)

# 5. List Operations
print("\n--- List Operations ---")
a = [1, 2, 3]
b = [4, 5, 6]
print("Concatenation:", a + b)
print("Repetition:", a * 3)
print("Length:", len(a))
print("Membership (2 in a):", 2 in a)

# 6. List Comprehension
print("\n--- List Comprehension ---")
squares = [x**2 for x in range(1, 6)]
print("Squares:", squares)

evens = [x for x in range(10) if x % 2 == 0]
print("Even numbers:", evens)

# 7. Sorting and Reversing
print("\n--- Sorting and Reversing ---")
unsorted = [3, 1, 4, 1, 5, 9, 2, 6]
print("Sorted:", sorted(unsorted))           # Returns new list
unsorted.sort()                               # Sorts in place
print("After in-place sort:", unsorted)
unsorted.reverse()
print("After reverse:", unsorted)

# 8. Useful Built-in Functions
print("\n--- Built-in Functions ---")
nums = [10, 20, 30, 40, 50]
print("Min:", min(nums))
print("Max:", max(nums))
print("Sum:", sum(nums))
print("Enumerate:", list(enumerate(nums)))
````

## Traversing a List with Index

> Uses `range(len(list))` to print each item with its index position.

````python
# Memory location demo using id() + mutability concept

# Mutable object (list): in-place change keeps same memory location
mutable_demo = numbers.copy()  # using existing `numbers` from notebook
print("Mutable list initial:", mutable_demo)
print("Memory (id):", id(mutable_demo), "Hex:", hex(id(mutable_demo)))

mutable_demo.append(99)  # in-place mutation
print("\nAfter append (mutated in place):", mutable_demo)
print("Memory (id):", id(mutable_demo), "Hex:", hex(id(mutable_demo)))

# Immutable object (int): reassignment creates a new object (new memory location)
immutable_demo = i  # using existing `i`
print("\nImmutable int initial:", immutable_demo)
print("Memory (id):", id(immutable_demo), "Hex:", hex(id(immutable_demo)))

immutable_demo = immutable_demo + 1  # new int object
print("\nAfter reassignment (+1):", immutable_demo)
print("Memory (id):", id(immutable_demo), "Hex:", hex(id(immutable_demo)))
````

### ⚠️ Common Gotchas with Lists

- **Index starts at 0**, not 1 — `list[0]` is the first item
- **`IndexError`** if you access an index that doesn't exist
- **`.remove(value)`** removes the *first* occurrence only
- **`.pop()`** without argument removes the *last* item
- **Lists are mutable** — changes affect the original list!

```python
# Wrong — index out of range
x = [1, 2, 3]
print(x[5])  # ❌ IndexError!

# Right
print(x[-1])  # ✅ Last item = 3
```

### 📝 Quick Revision
| Operation | Code |
|-----------|------|
| Create | `lst = [1, 2, 3]` |
| Access | `lst[0]`, `lst[-1]` |
| Add to end | `lst.append(x)` |
| Insert at pos | `lst.insert(i, x)` |
| Remove by value | `lst.remove(x)` |
| Remove by index | `lst.pop(i)` |
| Change item | `lst[i] = new_val` |
| Length | `len(lst)` |
| Check membership | `x in lst` |

## 🔴 Challenge — Student Grade Manager

**Scenario:** You're building a simple grade management system for a class.

**Tasks:**
1. **Remove** the duplicate score `45` (only the second occurrence)
2. **Add** a new student: `"Mia"` scored `82` — append to both lists
3. Print the **top 3 scores** using slicing after sorting (don't modify the original!)
4. Use **list comprehension** to get names of students who scored **above 75**
5. Print a **formatted report** using `enumerate`:
   ```
   Rank 1: Raj — 72
   Rank 2: Priya — 45
   ...
   ```
6. Print the **class average** (use `sum()` and `len()`)

**Hint for Task 4:** `zip(names, scores)` pairs both lists together 🤝

````python
scores = [72, 45, 88, 95, 60, 45, 33, 78, 91, 55]
names  = ["Raj", "Priya", "Ali", "Sara", "Tom",
          "Nina", "Kiran", "Leo", "Zara", "Arjun"]

# Task 1 — Remove second 45
# YOUR CODE HERE
# Step 1:  Find the index of duplicate
duplicateIndex = scores.index(45, scores.index(45)+1)
scores.pop(duplicateIndex)
names.pop(duplicateIndex)
print(f"After duplicate removed, Scores: {scores}, Names: {names}")
# Task 2 — Add Mia with score 82
# YOUR CODE HERE
names.append("Mia")
scores.append(82)
print(f"After adding mia, Scores: {scores}, Names: {names}")
# Task 3 — Top 3 scores (don't modify original)
# YOUR CODE HERE
top3scores = sorted(scores, reverse=True)[:3]
print(f"Top 3 scores: {top3scores}")
# Task 4 — Names scoring above 75
# YOUR CODE HERE
namesmorethan75 = []
for i in range(len(scores)):
    if scores[i] > 75:
        namesmorethan75.append(names[i])
        
print(f"Names who scored more than 75: {namesmorethan75}")

# Task 5 — Formatted report with enumerate
# YOUR CODE HERE

# Task 6 — Class average
# YOUR CODE HERE
averagescore = sum(scores)/len(scores)
print(f"Average Scores: {averagescore}")
````

### ✅ Solution (try it yourself first!)

````python
scores = [72, 45, 88, 95, 60, 45, 33, 78, 91, 55]
names  = ["Raj", "Priya", "Ali", "Sara", "Tom",
          "Nina", "Kiran", "Leo", "Zara", "Arjun"]

# Task 1 — Remove second 45 (pop by index to target the duplicate)
second_idx = scores.index(45, scores.index(45) + 1)  # find 2nd occurrence
scores.pop(second_idx)
names.pop(second_idx)

# Task 2 — Add Mia with score 82
names.append("Mia")
scores.append(82)

# Task 3 — Top 3 scores (sorted descending, sliced)
top3 = sorted(scores, reverse=True)[:3]
print("Top 3 scores:", top3)

# Task 4 — Names scoring above 75
high_scorers = [name for name, score in zip(names, scores) if score > 75]
print("High scorers (>75):", high_scorers)

# Task 5 — Formatted report
print("\n--- Class Report ---")
for rank, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"Rank {rank}: {name} — {score}")

# Task 6 — Class average
average = sum(scores) / len(scores)
print(f"\nClass Average: {average:.2f}")
````
