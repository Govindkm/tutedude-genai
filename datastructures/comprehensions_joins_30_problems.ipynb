## Section 1: 2-Table Join Comprehensions (Problems 1-10)

1. Build `student_city_map`: `{name: city}` from `students`.
2. Build `course_category_map`: `{course: category}` from `courses`.
3. Build `high_score_rows`: list of score rows where `score >= 85` from `exam_scores`.
4. Build `order_status_map`: `{order_id: status}` from `orders`.
5. Build `product_price_map`: `{product: price}` from `products`.
6. Build `delivered_orders`: list of order dicts where status is Delivered.
7. Build `retail_customers`: list of customer dicts where segment is Retail.
8. Join `students + enrollments`: list of tuples `(student_name, course_id)`.
9. Join `products + order_items`: list of tuples `(product_name, qty)`.
10. Join `orders + customers`: list of tuples `(order_id, customer_name, status)`.

```python
# YOUR SOLUTIONS: Problems 1-10
p1_student_city_map = {student.get('name'): student.get('city') for student in students}
p2_course_category_map = {course.get('course'): course.get('category') for course in courses}
p3_high_score_rows = [exam_score for exam_score in exam_scores if exam_score.get('score') >= 85]
p4_order_status_map = {order.get('order_id'): order.get('status') for order in orders }
p5_product_price_map = {product.get('product'): product.get('price') for product in products}
p6_delivered_orders = [order for order in orders if order['status'] == "Delivered"]
p7_retail_customers = [customer for customer in customers if customer['segment'] == 'Retail']
p8_student_course_ids = [(student['name'], enrollment['course_id']) for student in students for enrollment in enrollments if student['student_id'] == enrollment['student_id']]
p9_product_qty = [(product["product"], order["qty"]) for product in products for order in order_items if product['product_id'] == order['product_id']]
p10_order_customer_status = [(order['order_id'], customer['name'], order['status']) for customer in customers for order in orders if order['customer_id']==customer['customer_id']]

# Optional quick prints
# print(p8_student_course_ids)
print(p1_student_city_map)
print(p2_course_category_map)
print(p3_high_score_rows)
print(p4_order_status_map)
print(p5_product_price_map)
print(p6_delivered_orders)
print(p7_retail_customers)
print(p8_student_course_ids)
print(p9_product_qty)
print(p10_order_customer_status)
```

## Section 2: 3-Table Join Comprehensions (Problems 11-20)

11. Join `students + enrollments + courses`: tuples `(student_name, course_name)`.
12. Join `students + exam_scores + courses`: tuples `(student_name, course_name, score)`.
13. Build `passed_records`: same as above but only `score >= 75`.
14. Build `city_python_students`: student names from Chennai enrolled in Python.
15. Build `course_toppers`: `{course_name: top_score}`.
16. Build `customer_order_items`: tuples `(customer_name, order_id, product_name, qty)`.
17. Build `delivered_line_items`: only delivered order line items with customer and product names.
18. Build `returned_products`: product names that appear in returned orders.
19. Build `segment_product_pairs`: tuples `(segment, product_name)` for delivered orders.
20. Build `customer_total_qty`: `{customer_name: total_qty}` using order_items + orders + customers.

```python
# Code to list down headers of each table with table name and headers present in there
def print_table_headers(table_name, table_data):
    if table_data:
        headers = table_data[0].keys()
        print(f"{table_name} Headers: {', '.join(headers)}")
    else:
        print(f"{table_name} is empty.")
```

```python
print_table_headers("Students", students)
print_table_headers("Courses", courses)
print_table_headers("Enrollments", enrollments)
print_table_headers("Exam Scores", exam_scores)
print_table_headers("Customers", customers)
print_table_headers("Products", products)
print_table_headers("Orders", orders)
print_table_headers("Order Items", order_items)


# YOUR SOLUTIONS: Problems 11-20
p11_student_course = [
    (student["name"], course["course"])
    for student in students
    for enrollment in enrollments
    if enrollment["student_id"] == student["student_id"]
    for course in courses
    if course["course_id"] == enrollment["course_id"]
]
p12_student_course_score = [
    (s["name"], c["course"], e["score"])
    for s in students
    for e in exam_scores
    if e["student_id"] == s["student_id"]
    for c in courses
    if c["course_id"] == e["course_id"]
]
p13_passed_records = [
    (s["name"], c["course"], e["score"])
    for s in students
    for e in exam_scores
    if e["student_id"] == s["student_id"] and e["score"] >= 75
    for c in courses
    if c["course_id"] == e["course_id"]
]
p14_city_python_students = [
    s["name"]
    for s in students
    if s["city"] == "Chennai"
    for c in courses
    if c["course"] == "Python"
    for e in enrollments
    if e["student_id"] == s["student_id"] and e["course_id"] == c["course_id"]
]
p15_course_toppers = {
    course["course"]: max(
        score["score"]
        for score in exam_scores
        if score["course_id"] == course["course_id"]
    )
    for course in courses
}
p16_customer_order_items = [
    (customer["name"], order["order_id"], product["product"], item["qty"])
    for customer in customers
    for order in orders
    if order["customer_id"] == customer["customer_id"]
    for product in products
    for item in order_items
    if order["order_id"] == item["order_id"]
    and product["product_id"] == item["product_id"]
]
p17_delivered_line_items = [
    {
        "order": order["order_id"],
        "customer": customer["name"],
        "products": [
            product["product"]
            for product in products
            for item in order_items
            if product["product_id"] == item["product_id"]
            and item["order_id"] == order["order_id"]
        ],
    }
    for order in orders
    for customer in customers
    if order["customer_id"] == order["customer_id"] and order["status"] == "Delivered"
]
p18_returned_products = [
    product["product"]
    for product in products
    for item in order_items
    for order in orders
    if order["status"] == "Returned" and order["order_id"] == item["order_id"] and item["product_id"] == product["product_id"]
]
p19_segment_product_pairs = [
    (customer["segment"], product["product"])
    for customer in customers
    for product in products
    for order in orders
    for item in order_items
    if order["status"] == "Delivered"
    and order["order_id"] == item["order_id"]
    and product["product_id"] == item["product_id"] 
    and order["customer_id"] == customer["customer_id"]
]
p20_customer_total_qty = {
    customer["name"]: sum(item['qty'] 
        for item in order_items 
        for order in orders
        if item["order_id"] == order["order_id"]
        if customer["customer_id"] == order["customer_id"]
    )
    for customer in customers
}

# Optional quick prints
# print(p16_customer_order_items[:3])
print(p11_student_course)
print(p12_student_course_score)
print(p13_passed_records)
print(p14_city_python_students)
print(p15_course_toppers)
print(p16_customer_order_items)
print(p17_delivered_line_items)
print(p18_returned_products)
print(p19_segment_product_pairs)
print(p20_customer_total_qty)
```
