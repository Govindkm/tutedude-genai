## Debug Tips

- Build helper maps first (e.g., `id -> name`, `id -> dict`).
- Start from a flat list of joined tuples, then convert to dict/list outputs.
- Validate one problem at a time with small `print()` checks.
- Use `round(value, 2)` for averages and revenue.
