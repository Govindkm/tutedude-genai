# Mixed Comprehensions Practice (Lists + Dictionaries)

**Goal:** Practice writing list and dictionary comprehensions from plain-English problems.

Run each task cell after writing your solution.

```python
# Setup data for all tasks
students = ["Asha", "Ravi", "Mina", "Karan", "Nina"]
marks = [91, 56, 78, 34, 88]
prices = [120, 75, 200, 40, 95]
products = ["Pen", "Notebook", "Bag", "Eraser", "Bottle"]

print("Data loaded")
```

## 🟢 Beginner Problems

1. Create a list `passing_marks` with marks greater than or equal to 60.
2. Create a dictionary `student_marks` by combining `students` and `marks`.

```python
# YOUR SOLUTION (Beginner)
passing_marks = [mark for mark in marks if mark >= 60]
student_marks = {student: mark for (student, mark) in zip(students, marks)}

# Write your code below

print("passing_marks:", passing_marks)
print("student_marks:", student_marks)
```

```python
# Self-check (Beginner)
assert passing_marks == [91, 78, 88], "Check your filtering condition for passing_marks"
assert student_marks["Asha"] == 91 and student_marks["Karan"] == 34, "Check your zip() mapping"
print("Beginner checks passed")
```

## 🟡 Intermediate Problems

1. Create a list `discounted_prices` where every product price gets 10% discount (keep 2 decimals).
2. Create a dictionary `passed_students` containing only students with marks >= 60.
3. Create a list `labels` in this format: `"Asha: 91"` for every student.

```python
# YOUR SOLUTION (Intermediate)
discounted_prices = [price*0.9 for price in prices]
passed_students = {student: mark for (student, mark) in student_marks.items() if mark >= 60}
labels = [f"{student}: {mark}" for (student, mark) in student_marks.items()]

print("discounted_prices:", discounted_prices)
print("passed_students:", passed_students)
print("labels:", labels)
```

```python
# Self-check (Intermediate)
assert discounted_prices == [108.0, 67.5, 180.0, 36.0, 85.5], "Discount calculation is incorrect"
assert passed_students == {"Asha": 91, "Mina": 78, "Nina": 88}, "Check dict comprehension filter"
assert labels[0] == "Asha: 91" and labels[-1] == "Nina: 88", "Check labels format"
print("Intermediate checks passed")
```

## 🔴 Challenge Problem

**Scenario:** Build a mini result dashboard.

Create these using comprehensions:
1. `grade_report` dictionary with rule: A=90+, B=75-89, C=60-74, F=<60.
2. `topper_names` list containing names with grade A.
3. `fail_count` integer counting how many students got F.
4. `score_plus_bonus` dictionary where every score is increased by 5 (max cap at 100).

```python
# YOUR SOLUTION (Challenge)
grade_report = {
    student: ("A" if score >= 90 else "B" if score >= 75 else "C" if score >= 60 else "F") 
    for student, score in student_marks.items()
}
topper_names = [name for name, grade in grade_report.items() if grade == 'A']
fail_count = list(grade_report.values()).count('F')
score_plus_bonus = {student: score+5 for student, score in student_marks.items()}

print("grade_report:", grade_report)
print("topper_names:", topper_names)
print("fail_count:", fail_count)
print("score_plus_bonus:", score_plus_bonus)
```

```python
# Self-check (Challenge)
assert grade_report == {"Asha": "A", "Ravi": "F", "Mina": "B", "Karan": "F", "Nina": "B"}, "Grade mapping is incorrect"
assert topper_names == ["Asha"], "Topper extraction is incorrect"
assert fail_count == 2, "Fail count should be 2"
assert score_plus_bonus == {"Asha": 96, "Ravi": 61, "Mina": 83, "Karan": 39, "Nina": 93}, "Bonus mapping is incorrect"
print("Challenge checks passed")
```
