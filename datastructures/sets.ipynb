## Sets in Python

**Analogy:** A set is like a container where all items are unique — if you add the same item twice, it stays just once.

**What it is:** A set is an unordered, mutable collection of unique values.

### Key Differences from Tuples & Lists

| Feature | List | Tuple | Set |
|---------|------|-------|-----|
| Ordered | ✅ | ✅ | ❌ No order |
| Mutable (changeable) | ✅ | ❌ | ✅ |
| Allows duplicates | ✅ | ✅ | ❌ No duplicates |
| Use `in` for membership | ✅ | ✅ | ✅ (faster!) |
| Syntax | `[]` | `()` | `{}` |

**When to use Sets:**
- Remove duplicates from a list
- Check if value exists (super fast!)
- Find common items between collections (intersection)
- Find unique items between collections (difference)

### Live Demo: Creating Sets & Removing Duplicates

```python
# Minimal working example - run this first!

# Creating a set - notice: curly braces {}
fruits = {"apple", "banana", "mango", "apple"}  # apple appears twice!
print("Set:", fruits)  # But in output, apple only appears once
print("Type:", type(fruits))
print("Length:", len(fruits))

# Converting a list with duplicates to a set
numbers_list = [1, 2, 3, 2, 4, 3, 5]
unique_numbers = set(numbers_list)
print("\nList with duplicates:", numbers_list)
print("Converted to set:", unique_numbers)

# Membership check (is this item in the set?)
print("\nIs 'banana' in fruits?", "banana" in fruits)
print("Is 'grape' in fruits?", "grape" in fruits)
```

### Live Demo: Set Operations (Union, Intersection, Difference)

```python
# Set operations are like Venn diagrams!

# Create two sets
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("Set A:", set_a)
print("Set B:", set_b)
print()

# 1. UNION: All items from both sets (no duplicates)
union_result = set_a | set_b  # or set_a.union(set_b)
print("A | B (Union):", union_result)
print("\t➜ All unique items from A and B")

# 2. INTERSECTION: Items common to both
intersection_result = set_a & set_b  # or set_a.intersection(set_b)
print("\nA & B (Intersection):", intersection_result)
print("\t➜ Only items that appear in BOTH sets")

# 3. DIFFERENCE: Items in A but not in B
difference_result = set_a - set_b  # or set_a.difference(set_b)
print("\nA - B (Difference):", difference_result)
print("\t➜ Items in A that are NOT in B")

# 4. SYMMETRIC DIFFERENCE: Items in A or B, but not both
sym_diff = set_a ^ set_b  # or set_a.symmetric_difference(set_b)
print("\nA ^ B (Symmetric Difference):", sym_diff)
print("\t➜ Items that are unique to A OR unique to B (but not in both)")
```

### Interactive Challenge: Set Operations

```python
# 🟢 Beginner: Union & Intersection
hobbies_asha = {"reading", "coding", "sports"}
hobbies_ravi = {"coding", "gaming", "sports"}
# Find common hobbies and all hobbies
print(f"Asha's hobbies: {hobbies_asha}, Ravi's hobbies: {hobbies_ravi}\nCommon Hobbies: {hobbies_ravi | hobbies_asha}")

# 🟡 Intermediate: Difference
programming_langs_learned = {"Python", "JavaScript", "Java"}
programming_langs_interested = {"Python", "Go", "Rust", "C++"}
# Languages interested in but haven't learned yet = interested - learned
print(f"Learned: {programming_langs_learned}, Interested: {programming_langs_interested}\nStill to learn: {programming_langs_interested - programming_langs_learned}")

# 🔴 Challenge: Multi-step filtering
class_a_roll_numbers = {101, 102, 103, 104, 105}
class_b_roll_numbers = {104, 105, 106, 107, 108}
absent_today = {102, 107}
# Students present today from both classes
present_today = (class_a_roll_numbers | class_b_roll_numbers) - absent_today
print(present_today)
```

### Adding & Removing Items from Sets

```python
# Sets are MUTABLE - you can add and remove items

fruits = {"apple", "banana"}
print("Initial set:", fruits)

# ADD an item
fruits.add("mango")
print("After add('mango'):", fruits)

# Try adding a duplicate
fruits.add("apple")  # Already exists, so nothing changes
print("After add('apple'):", fruits)

# REMOVE an item (raises error if not found)
fruits.remove("banana")
print("After remove('banana'):", fruits)

# DISCARD an item (doesn't raise error if not found)
fruits.discard("grape")  # Doesn't exist, but no error
print("After discard('grape'):", fruits)

# CLEAR the entire set
fruits_copy = {"x", "y", "z"}
fruits_copy.clear()
print("After clear():", fruits_copy)
```

### ⚠️ Watch Out For

- **Empty set syntax:** Use `set()`, NOT `{}` (which creates an empty dict)
- **No indexing:** Sets are unordered, so `s[0]` raises an error
- **No duplicates allowed:** If you add a duplicate, it's silently ignored
- **`remove()` vs `discard()`:** 
  - `remove()` raises error if item not found
  - `discard()` silently ignores if item not found

### 📝 Quick Revision

| What | How | Syntax |
|------|-----|--------|
| Create set | Use `{}` or `set(list)` | `s = {1, 2, 3}` |
| Empty set | Use `set()` not `{}` | `s = set()` |
| Union (all items) | `|` or `.union()` | `a \| b` |
| Intersection (common) | `&` or `.intersection()` | `a & b` |
| Difference (in A not B) | `-` or `.difference()` | `a - b` |
| Add item | `.add()` | `s.add(5)` |
| Remove item | `.remove()` (errors) or `.discard()` | `s.remove(5)` |
| Check membership | `in` | `5 in s` |

**Use sets when:** You need unique values, want lightning-fast membership checks, or need to find common/unique items between collections.
