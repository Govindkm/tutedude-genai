## Tuples in Python

**Analogy:** A tuple is like a sealed lunch box. You can open it and see what is inside, but you cannot replace items after sealing.

**What it is:** A tuple is an ordered, immutable collection in Python.

````python
# Minimal working example - run this first!
fruits = ("apple", "banana", "mango")
print("Tuple:", fruits)
print("Type:", type(fruits))
print("First item:", fruits[0])
print("Length:", len(fruits))

# Unpacking
a, b, c = fruits
print("Unpacked values:", a, b, c)
````

### Watch out for

- Single-item tuple needs a comma: `(5,)`
- Parentheses are optional in many cases, commas create the tuple: `x = 1, 2, 3`
- Tuples are immutable, so `t[0] = 10` raises `TypeError`
- To add items, create a new tuple: `t = t + (new_item,)`

### Quick Revision

| What | How |
|------|-----|
| Tuple | Ordered, immutable collection |
| Access | Use indexing like `t[0]` |
| Unpack | `a, b = t` (matching number of values) |
| Add item | Create a new tuple with concatenation |

**Use tuples when:** data should stay fixed, such as coordinates, menu options, or constant settings.

### Next Practice: Slicing, Membership, and Nested Tuples

**Analogy:** If a tuple is a train, slicing lets you pick a set of coaches, and nested tuples are like coaches with smaller cabins inside.

**What to do:** Run the next code cell and complete each task.

- 🟢 Slicing: Extract first 3 and last 2 items from a tuple.
- 🟡 Membership: Check if a value exists using `in`.
- 🔴 Nested Tuple: Access values from inner tuples and compute a total.

````python
# Run and complete this practice step by step

# Given data
numbers = (10, 20, 30, 40, 50)

# 🟢 Slicing practice
first_three = numbers[:3]
last_two = numbers[-2:]
print("First three:", first_three)
print("Last two:", last_two)

# 🟡 Membership practice
check_value = 30
print(f"Is {check_value} in numbers?", check_value in numbers)

# 🔴 Nested tuple practice
students = (
    ("Asha", (85, 90, 88)),
    ("Ravi", (78, 82, 80, 100))
)

# Access Asha's second mark and Ravi's total marks
asha_second_mark = students[0][1][1]
ravi_total = sum(students[1][1])

print("Asha second mark:", asha_second_mark)
print("Ravi total:", ravi_total)

# Your mini challenge:
# 1) Print only the names as a tuple
names = tuple([student[0] for student in students])
print(names)
# 2) Print average marks for Asha
score = [student[1] for student in students if student[0] == "Asha"]
avg = sum(score[0])/len(score[0])
print(avg)
# 3) Check if 100 is present in Ravi's marks
present = 100 in [student[1] for student in students if student[0] == "Ravi"][0]
print(present)
````
