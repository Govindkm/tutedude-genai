# Data Structures in Python

Data Structures in Python are being used for storing the data in a structured format so that we can access the data and alon with that we can perform all of the operations with these data structures that we can perform in the case of databases as well like CRUD operations(create, read, update, delete)

## Types of Data Structures in Python

> These are the most commonly used built-in data structures in Python.
> Each one is useful in different situations.

## 1. List
A **list** stores multiple items in a single variable, in an **ordered** way.

**Basic points:**
- Lists are **mutable** (you can change, add, or remove items).
- They allow **duplicate values**.
- Items are accessed by **index** (starting from `0`).

**Example:**
```python
fruits = ["apple", "banana", "apple"]
fruits.append("mango")
print(fruits[1])  # banana
```

**Use list when:**
- Order matters.
- You need to update data frequently.

## 2. Tuple
A **tuple** is similar to a list, but it is **immutable** (cannot be changed after creation).

**Basic points:**
- Tuples are **ordered**.
- They allow **duplicate values**.
- Better for fixed data that should not change.

**Example:**
```python
coordinates = (10, 20)
print(coordinates[0])  # 10
```

**Use tuple when:**
- Data is fixed (like coordinates, days of week).
- You want safer, read-only grouped values.

## 3. Set
A **set** stores **unique** items only.

**Basic points:**
- Sets are **unordered** (no index positions).
- Duplicate values are automatically removed.
- Useful for membership checks and removing duplicates.

**Example:**
```python
numbers = {1, 2, 2, 3}
print(numbers)  # {1, 2, 3}
```

**Use set when:**
- You need only unique values.
- You want fast `in` checks.

## 4. Dictionary
A **dictionary** stores data in **key-value pairs**.

**Basic points:**
- Keys are unique.
- Values can be of any data type.
- Very useful for mapping related information.

**Example:**
```python
student = {"name": "Govin", "age": 20, "course": "Python"}
print(student["name"])  # Govin
```

**Use dictionary when:**
- You want to access data using a meaningful key (not index).
- You are storing labeled data (like user profiles, records).
