# Dictionaries in Python

> A dictionary is an **unordered, mutable** collection of **key-value pairs**.
> Think of it like a contact book — you look up a **name (key)** to get a **phone number (value)**.

```
contact_book = {
    "Alice" : "9876543210",
    "Bob"   : "9123456789"
}
```

| Feature | List | Dictionary |
|---------|------|------------|
| Access by | Index (0, 1, 2...) | Key (any immutable value) |
| Ordered | ✅ Yes | ✅ Yes (Python 3.7+) |
| Mutable | ✅ Yes | ✅ Yes |
| Duplicates | ✅ Allowed | ❌ Keys must be unique |

````python
# ── 1. Creating Dictionaries ──────────────────────────────────────────

# Empty dict
empty = {}

# Student record
student = {
    "name": "Alice",
    "age": 20,
    "score": 85,
    "grade": "A"
}

# ── 2. Accessing values ───────────────────────────────────────────────
print(student["name"])           # Alice  — direct access
print(student.get("age"))        # 20     — safe access (returns None if key missing)
print(student.get("gpa", 0.0))   # 0.0   — default value if key not found

# ── 3. Adding & Updating ──────────────────────────────────────────────
student["age"] = 21              # update existing key
student["city"] = "Mumbai"       # add new key
print(student)

# ── 4. Deleting ───────────────────────────────────────────────────────
del student["city"]              # delete a key
popped = student.pop("grade")    # remove and return value
print(f"Popped: {popped}")
print(student)
````

## 🔑 Keys, Values, Items

Three essential views of a dictionary:

````python
inventory = {
    "apples": 50,
    "bananas": 30,
    "oranges": 20
}

print(inventory.keys())    # dict_keys(['apples', 'bananas', 'oranges'])
print(inventory.values())  # dict_values([50, 30, 20])
print(inventory.items())   # dict_items([('apples', 50), ('bananas', 30), ('oranges', 20)])

# ── Looping ───────────────────────────────────────────────────────────
print("\n--- Loop over keys ---")
for item in inventory:                         # iterates keys by default
    print(item)

print("\n--- Loop over key-value pairs ---")
for item, qty in inventory.items():            # unpack key and value
    print(f"{item}: {qty} units")

# ── Check membership ──────────────────────────────────────────────────
print("\napples" in inventory)    # True  — checks keys
print(50 in inventory.values())  # True  — checks values
````

## 🔧 Useful Dictionary Methods

| Method | What it does |
|--------|--------------|
| `d.get(key, default)` | Safe access — returns default if key missing |
| `d.keys()` | All keys |
| `d.values()` | All values |
| `d.items()` | All key-value pairs as tuples |
| `d.pop(key)` | Remove & return value for key |
| `d.update({...})` | Merge another dict into `d` |
| `d.copy()` | Shallow copy |
| `d.clear()` | Remove all items |

````python
# ── update() — merge dictionaries ────────────────────────────────────
base = {"name": "Bob", "score": 70}
update = {"score": 88, "grade": "B+"}   # score will be overwritten

base.update(update)
print(base)   # {'name': 'Bob', 'score': 88, 'grade': 'B+'}

# ── Dictionary Comprehension ──────────────────────────────────────────
# Same idea as list comprehension, but produces {key: value} pairs

names  = ["Alice", "Bob", "Charlie"]
scores = [85, 72, 91]

# Build a dict from two lists
gradebook = {name: score for name, score in zip(names, scores)}
print(gradebook)   # {'Alice': 85, 'Bob': 72, 'Charlie': 91}

# Filter — only students who passed (score >= 75)
passed = {name: score for name, score in gradebook.items() if score >= 75}
print(passed)   # {'Alice': 85, 'Charlie': 91}
````

## ⚠️ Common Gotchas

**1. KeyError — accessing a missing key directly**
```python
d = {"a": 1}
print(d["b"])          # ❌ KeyError!
print(d.get("b", 0))   # ✅ Returns 0 safely
```

**2. Keys must be immutable**
```python
d = {[1, 2]: "value"}  # ❌ TypeError — lists can't be keys
d = {(1, 2): "value"}  # ✅ Tuples are fine
```

**3. Duplicate keys — last one wins**
```python
d = {"a": 1, "a": 99}  # No error, but d["a"] == 99
```

## 📝 Quick Revision

| What | How |
|------|-----|
| Create | `d = {"key": value}` |
| Access (safe) | `d.get("key", default)` |
| Add / Update | `d["key"] = value` |
| Delete | `del d["key"]` or `d.pop("key")` |
| Loop key+value | `for k, v in d.items():` |
| Comprehension | `{k: v for k, v in ... if cond}` |
| Top value | `max(d, key=lambda k: d[k])` |
