# Python Inheritance

**Analogy:** Inheritance is like a specialized version of a general blueprint. A child class receives common features from a parent class and can add or customize its own features.

**What it is:** Inheritance lets one class reuse and extend another class.

We will build this topic progressively: base classes, child classes, overriding, `super()`, polymorphism, multiple inheritance, and a small project.

## 1. Set Up Notebook and Base Class

A base class contains behavior shared by related objects. Start with a simple `Person` class and verify its normal behavior.

```python
class Person:
    species = "human"

    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"My name is {self.name}."


person = Person("Govind")
print(person.name)
print(person.species)
print(person.introduce())
```

## 2. Create Single-Inheritance Child Classes

A child class is declared with the parent class in parentheses. It automatically receives the parent's attributes and methods, and it can add new behavior.

```python
class Student(Person):
    def study(self):
        return f"{self.name} is studying."


class Employee(Person):
    def work(self):
        return f"{self.name} is working."


student = Student("Govind")
employee = Employee("Asha")

print(student.introduce())
print(student.study())
print(employee.introduce())
print(employee.work())
```

## 3. Override Methods in Subclasses

A child class can replace a parent method with a specialized implementation. The method name stays the same, but the behavior changes.

```python
class Student(Person):
    def introduce(self):
        return f"I am {self.name}, a student."


class Employee(Person):
    def introduce(self):
        return f"I am {self.name}, an employee."


people = [Student("Govind"), Employee("Asha")]

for person in people:
    print(person.introduce())
```

## 4. Reuse Parent Initialization with `super()`

When a child has its own `__init__`, call `super().__init__(...)` to reuse the parent's setup instead of repeating it.

```python
class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def study(self):
        return f"{self.name} is studying {self.course}."


student = Student("Govind", "Python")
print(student.introduce())
print(student.study())
```

## 5. Use Polymorphism with a Shared Interface

**Polymorphism** means different object types can respond to the same method call in their own way. The function below only needs each object to provide `introduce()`.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"My name is {self.name}."


class Student(Person):
    def introduce(self):
        return f"{self.name} is a student."


class Employee(Person):
    def introduce(self):
        return f"{self.name} is an employee."


def introduce_everyone(people):
    for person in people:
        print(person.introduce())


introduce_everyone([
    Person("Ravi"),
    Student("Govind"),
    Employee("Asha"),
])
```

## 6. Multiple Inheritance and MRO

Multiple inheritance lets a class inherit from more than one parent. Python uses the **method resolution order (MRO)** to decide where to search first when a method exists in multiple classes.

```python
class Camera:
    def take_photo(self):
        return "Photo taken"


class Phone:
    def make_call(self):
        return "Calling..."


class SmartPhone(Camera, Phone):
    pass


smartphone = SmartPhone()
print(smartphone.take_photo())
print(smartphone.make_call())
print(SmartPhone.__mro__)
```

## 7. Small Inheritance-Based Mini Project

We will model different employee roles. Every employee can report, while each role customizes its own work description.

```python
class Employee:
    company = "TuteDude"

    def __init__(self, name):
        self.name = name

    def report(self):
        return f"{self.name} works at {self.company}."

    def work(self):
        raise NotImplementedError("Child classes must define work().")


class Developer(Employee):
    def work(self):
        return f"{self.name} writes code."


class Designer(Employee):
    def work(self):
        return f"{self.name} designs interfaces."


def print_employee_summary(employees):
    for employee in employees:
        print(employee.report())
        print(employee.work())


print_employee_summary([
    Developer("Govind"),
    Designer("Asha"),
])
```

## 8. Quick Validation with Assertions

Assertions turn expected behavior into automatic checks. If one condition is wrong, Python raises an `AssertionError`.

```python
developer = Developer("Govind")
designer = Designer("Asha")

assert isinstance(developer, Employee)
assert isinstance(developer, Developer)
assert developer.company == "TuteDude"
assert developer.report() == "Govind works at TuteDude."
assert developer.work() == "Govind writes code."
assert designer.work() == "Asha designs interfaces."

print("All inheritance checks passed.")
```
