## Class Methods and Static Methods

### Class Method

A class method is a method that receives the **class itself** as its first argument, written as `cls`.

```python
@classmethod
def school_name(cls):
    return cls.school
```

Because `cls` refers to the class, the method can read or change class attributes. For example, `Student.school_name()` returns `"TuteDude"` by reading `Student.school`.

### Why does Python have class methods?

Different OOP languages use different names and features. Many languages commonly teach instance methods and static methods, but Python adds class methods because they solve a useful middle-ground problem:

- An **instance method** works with one object through `self`.
- A **class method** works with the class through `cls` and can also work correctly with subclasses.
- A **static method** is a helper grouped inside a class; it receives neither `self` nor `cls`.

Class methods are useful for:

- Reading or updating shared class-level data
- Creating alternative ways to construct objects, called alternative constructors
- Writing methods that should use the actual subclass when inheritance is involved

A static method could read `Student.school` directly, but it would not automatically receive the class as `cls`. A class method makes that relationship explicit and supports inheritance more naturally.

### Method Comparison

| Method | First argument | Works mainly with |
|---|---|---|
| Instance method | `self` | One object |
| Class method | `cls` | The class and shared class data |
| Static method | None | Independent helper logic |


````python
class Student:
    school = "TuteDude"
    total_students = 0

    def __init__(self, name, course):
        self.name = name
        self.course = course
        Student.total_students += 1

    @classmethod
    def school_name(cls):
        return cls.school

    @staticmethod
    def is_valid_age(age):
        return age >= 18

    @classmethod
    def show_total_students(cls):
        return cls.total_students
    
    @staticmethod
    def is_valid_course(course):
        return len(course)!=0

student = Student("Govind", "Python")

print(student.school_name())
print(Student.school_name())
print(Student.is_valid_age(25))
print(Student.is_valid_age(15))
print(Student.show_total_students())
print(Student.is_valid_course("Python"))
print(Student.is_valid_course(""))
````

### Why use `cls` instead of the class name?

This version is less flexible:

```python
@classmethod
def school_name(cls):
    return Student.school
```

This version is preferred:

```python
@classmethod
def school_name(cls):
    return cls.school
```

`cls` means “the class that called this method.” If a child class inherits `school_name()`, `cls` can refer to the child class. This makes the method work naturally with inheritance.

### Alternative Constructors

A class method can create an object in another format. This is called an alternative constructor:

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    @classmethod
    def from_text(cls, text):
        name, course = text.split(",")
        return cls(name.strip(), course.strip())


student = Student.from_text("Govind, Python")
print(student.name)
print(student.course)
```

`from_text()` receives text, converts it into the values required by `__init__`, and returns a new `Student` object.


````python
class Course:
    total_courses = 0

    def __init__(self, name, duration):
        self.name = name
        self.duration = duration
        Course.total_courses += 1

    @classmethod
    def show_total_courses(cls):
        return cls.total_courses

    @classmethod
    def from_text(cls, text):
        name, duration = text.split(",")
        return cls(name.strip(), duration.strip())

    @staticmethod
    def is_valid_duration(duration):
        return duration.strip() != ""


course_one = Course("Python", "8 weeks")
course_two = Course.from_text("SQL, 6 weeks")

print(course_one.name)
print(course_two.name)
print(Course.show_total_courses())
print(Course.is_valid_duration("4 weeks"))
print(Course.is_valid_duration(""))
````

### Common Gotcha

A class method should use `cls` when it needs class-level information or when it creates an object:

```python
return cls(...)
```

A static method should not be forced to use `cls`. Use `@staticmethod` when the logic is related to the class but does not need object or class data.


````python
class User:
    
    domain = "example.com"
    
    def __init__(self, name, email):
        self.name = name
        self.email = email
        
    @classmethod
    def from_email(cls, email:str):
        name, mail = email.split("@")
        return cls(name, email)
    
    @staticmethod
    def is_valid_email(email:str):
        return email.find("@")>0
````

````python
user = User.from_email("govind@example.com")
print(user.name)
print(User.is_valid_email("govind@example.com"))
print(User.is_valid_email("invalid-email"))
````
