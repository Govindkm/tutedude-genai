## Object Oriented Programming in Python

**Analogy:** A class is like a blueprint for making many similar things. An object is one real thing made from that blueprint.

**What it is:** OOP helps you organize code by grouping data and behavior together inside classes.

```python
# Minimal working example
class Dog:
    def bark(self):
        return "Woof!"

dog1 = Dog()
print(dog1.bark())
```

### Your Turn

**Problem:** Create a class `Car` with:
- an `__init__` method that stores `brand` and `model`
- a method `details()` that returns `Brand: X, Model: Y`

Create one car object and print its details.

```python
# YOUR CODE HERE
# 1) define Car
# 2) create one object
# 3) print details

class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    def details(self):
        return f"Brand: {self.brand}, Model: {self.model}"
    
myCar = Car("Maruti", "Baleno")

print(myCar.details())
```

### Watch out for
- `self` refers to the current object.
- `__init__` runs automatically when you create an object.
- Methods inside a class need `self` as the first parameter.

### Quick Revision
| What | How |
|------|-----|
| Class | Blueprint for objects |
| Object | Real instance of a class |
| `__init__` | Sets up object data |
| Method | Function inside a class |

```python
# Create a class Student
# 1. __init__ should store name and marks
# 2. create a method result() that returns "Pass" if marks >= 40 else "Fail"
# 3. create 2 student objects and print their name + result

class Student:
    # YOUR CODE HERE
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        return "Pass" if self.marks >= 40 else "Fail"
    
student1 = Student("Govind", 87)
student2 = Student("Ajit", 65)

print(f"Name: {student1.name}, Result: {student1.result()}")

print(f"Name: {student2.name}, Result: {student2.result()}")
```
