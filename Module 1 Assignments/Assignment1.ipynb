# Task 1: Product Collections(Lists & Tuples)

```python
products = ["keyboard", "banana", "toothpaste", "potatoes", "chair", "led bulb"]
sample_product = (products[0], 499, "electronics")

print(f"2nd product: {products[1]}, last product: {products[-1]}")

# Appending products to the list
products.append("table")
products.append("monitor")

print(f"Updated products list: {products}")

try:
    sample_product[1] = 599
except TypeError as e:
    print(f"Error: {e} - Tuples are immutable, cannot modify elements.")
    
# Update sample_product by converting to a list, modifying, and converting back to a tuple
sample_product_list = list(sample_product)
print(f"Original sample_product: {sample_product}")
sample_product_list[1] = 599
sample_product = tuple(sample_product_list)
print(f"Updated sample_product: {sample_product}")
```

# Task 2: Categories (Sets)

```python
products = ["keyboard", "banana", "toothpaste", "potatoes", "chair", "led bulb"]
categories = ["electronics", "food", "personal care", "food", "furniture", "electronics"]

categories_set = set(categories)
print(f"Categories set: {categories_set}, size: {len(categories_set)}")

# Add a new category to the set
categories_set.add("home decor")
print(f"Updated categories set: {categories_set}, size: {len(categories_set)}")
# Add a duplicate category to the set
categories_set.add("electronics")
print(f"Categories set after adding duplicate: {categories_set}, size: {len(categories_set)}")

# Check whether a category exists in the set
category_to_check = "food"
print(f"Checking if '{category_to_check}' is in categories set: {category_to_check in categories_set}")

print(f"Total number of categories: {len(categories)}, unique categories: {len(set(categories))}")
```

## Task 3: Product Pricing (Dictionaries)

1. Create a dictionary price_dict where keys are product names and values are prices(integers or floats). Include at least 6 entries.
2. Write small code blocks to:
    - Add a new product with price to price_dict.
    - Update the price of an existing product.
    - Remove a product by name(handle the case when the product does not exist).
3. Print the average price of all products(use only dictiionary operators and basic arithmetic)
Extra(optional): Print the product with both the maximum and minimum prices.


```python
import random as rnd
## 1
price_dict = {key: rnd.randint(1, 1000) for key in products}
print(price_dict)
price_dict["mousepad"] = 230
print(price_dict)
price_dict["toothpaste"] = 34
print(price_dict)
price_dict.pop("banana", None)
print(price_dict)
average = sum([value for value in price_dict.values()])/len(price_dict)
print(average)
max_price, min_price = (max([value for value in price_dict.values()]),min([value for value in price_dict.values()])) 
max_priced_product = [product for product in price_dict.items() if product[1] == max_price]
min_priced_product = [product for product in price_dict.items() if product[1] == min_price]
print(max_priced_product[0])
print(min_priced_product[0])
```

## Task 4: Combined Operations

1. Using the products list and price_dict, create a list of tupes named catalog where each tuple is (product_name, price, category).
2. From catalog, create a new dictionary category_to_products that maps each category to a list of product names in that category.
3. Print all products that belong to the category that has the maximum number of Products. 

```python
products = ["keyboard", "banana", "toothpaste", "potatoes", "chair", "led bulb"]
categories = ["electronics", "food", "personal care", "food", "furniture", "electronics"]
price_dict = {key: rnd.randint(1, 1000) for key in products}
categories_set = set(categories)
catelog = []
for i in range(len(products)):
    catelog.append((products[i], price_dict[products[i]], categories[i]))
    
print(catelog)

category_to_product = {category: [product[0] for product in catelog if product[2] == category] for category in categories_set}
print(category_to_product)

max_products = []
for item in category_to_product.items():
    if(len(item[1])>len(max_products)):
        max_products = item[1]
        
print(max_products)
```
