## Task 1: Discount Rules (if/elif/else)
1. Write a program that reads an integer `order_amount` from the user using `input()`.
2. Apply the following discount rules and print the final amount:
   - `order_amount < 2000` -> 0% discount
   - `2000 <= order_amount < 5000` -> 5% discount
   - `order_amount >= 5000` -> 10% discount
3. Ensure input is converted to a number and handle non-numeric input by displaying an error message and exiting the program.

```python
# Task 1 solution
# TODO: Read order amount from user
try: 
    order_amount = float(input("Enter order amount"))
# TODO: Apply discount with if/elif/else
    if order_amount >= 2000 and order_amount < 5000:
       order_amount = (1 - 0.05) * order_amount
    elif order_amount >= 5000:
       order_amount = (1 - 0.1) * order_amount 
# TODO: Print discount and final amount
    print(f"Price after discount: {order_amount}")
# TODO: Handle invalid (non-numeric) input
except:
    print("Please enter a numeric value")
```

## Task 2: Process Multiple Orders (for loop)
1. Given a list of order amounts `[1200, 2500, 880, 17500, 3000]`, use a `for` loop to:
   - Apply discount rules from Task 1 for each amount
   - Print a summary table showing `order_amount`, `discount`, and `final_amount`
2. Compute and print total revenue after discounts.

**Extra (optional):** Print the number of orders that received a discount (`discount > 0`).

```python
# Task 2 solution
orders = [1200, 2500, 880, 17500, 3000]

# TODO: Loop through orders
discount = [1 if price < 2000 else 0.95 if price<5000 else 0.9 for price in orders]
discounted = [price*discount for price, discount in zip(orders, discount)]
summary = zip(orders, discount, discounted)
print(f"Discounted price: {discounted}")
# TODO: Apply discount rules
# TODO: Print summary rows
print(f"Price|Discount|Final Price")
for item in summary:
    print(f"{item[0]}|{item[1]}|{item[2]}")
# TODO: Compute total revenue after discounts
total_revenue = sum(discounted)
print(f"Total revenue after selling: {total_revenue}")
# TODO (optional): Count discounted orders
count_discounted = [discount_item for discount_item in discount if discount_item<1]
print(f"Total number of discounted products are: {len(count_discounted)}")
```

## Task 3: User Menu (while loop + break/continue)

## Task 4: Data Validation and Skip Bad Data (for + continue + break)
Given the list values:
`[1200, 2500, -5, 3200, "error", 4500, 0, 1800]`

1. Iterate through the list and:
   - If a value is negative, treat it as corrupted data and `break` the loop
   - If a value is non-numeric, `continue` and skip adding it to sales
   - For valid values, add to `total_sales` and print the running total
2. Print the final total after the loop completes (or stops due to corrupted data).

```python
# Task 4 solution
values = [1200, 2500, 3200, "error", -5, 4500, 0, 1800]
total_sales = 0
correct_values = []
for item in values:
    if type(item) == str:
        continue
    if item<0:
        break
    correct_values.append(item)
    total_sales = total_sales + item
    print(f"Running total: {total_sales}")

print(total_sales)

# TODO: Loop and validate values
# TODO: break on negative values
# TODO: continue on non-numeric values
# TODO: Add valid values and print running total
# TODO: Print final total
```
