## Task 1: Write Sales Records to a File
1. Create a list of sales amounts:
   `sales = [1200, 450, 980, 1500, 3000]`
2. Open a file named `sales_data.txt` in write mode.
3. Write each sale on a new line in the file.
4. Close the file, then reopen it in read mode.
5. Print the file contents.

**Extra (optional):** Write the data in comma-separated format instead of separate lines.

````python
# Task 1 solution
sales = [1200, 450, 980, 1500, 3000]

# TODO: Write each sale to sales_data.txt (one per line)
with open("sales_data.txt", "w") as file:
    for record in sales:
        file.writelines(f"{record}\n")

# TODO: Reopen and print file contents
with open("sales_data.txt", "r") as file:
    data = file.readlines()
    print([int(line) for line in data])
    
````

## Task 2: Read File in Different Ways
Using `sales_data.txt`:
1. Read the entire file using `read()` and print it.
2. Read line by line using `readline()`.
3. Read all lines using `readlines()`, then convert them into a list of integers.
4. Ensure proper formatting by removing newline characters.

````python
# Task 2 solution
# TODO: Read full file using read()
with open("sales_data.txt", "r") as file:
    data = file.read()
    print(data)

# TODO: Read line by line using readline()
with open("sales_data.txt", "r") as file:
    while(True):
        line = file.readline()
        if(len(line)>0):
            print(line)
        else:
            break

# TODO: Use readlines() and convert to list[int]
sales_numbers = []
with open("sales_data.txt", "r") as file:
    lines = file.readlines()
    sales_numbers = [int(line) for line in lines]

print(sales_numbers)
````

## Task 3: Append New Sales
1. Append these new sales to the same file:
   - `500`
   - `750`
   - `1100`
2. After appending, reopen and print the updated file.

**Extra (optional):** Print the total number of lines after appending.

````python
# Task 3 solution
new_sales = [500, 750, 1100]

# TODO: Append new sales to sales_data.txt
with open("sales_data.txt", "a") as file:
    for item in new_sales:
        file.write(f"{item}\n")

# TODO: Reopen and print updated file
with open("sales_data.txt", "r") as file:
    lines = file.readlines()
    lines = list(map(lambda line: int(line), lines))
    print(lines)

# TODO (optional): Print total number of lines
with open("sales_data.txt", "r") as file:
    lines = file.readlines()
    print(f"Total no of lines: {len(lines)}")
````

## Task 4: Generate Summary Report from File
Using only file reading operations:
1. Read all sales values from `sales_data.txt`.
2. Convert each value to `int`.
3. Calculate and print:
   - Total Sales
   - Highest Sale
   - Lowest Sale
   - Average Sale

````python
# Task 4 solution
sales_values = []

# TODO: Read and convert all lines to integers
with open("sales_data.txt", "r") as file:
    lines = file.readlines()
    sales_values = list(map(lambda line: int(line), lines))
    total = sum(sales_values)
    highest = max(sales_values)
    lowest = min(sales_values)
    avg = total/len(sales_values)
    print(f"Total Sales: {total}, Highest: {highest}, Lowest: {lowest}, Average: {avg}")

# TODO: Calculate and print total, highest, lowest, and average
````

## Task 5: Create Product Info File (User Input)
1. Ask the user for 3 product names and their prices.
2. Save entries to `products.txt` in this format:
   `Product | Price`
3. Reopen `products.txt` and print each line with clear formatting.


````python
# Task 5 solution
# TODO: Collect 3 product names and prices from input
# TODO: Save as Product | Price in products.txt
with open("products.txt", "w") as file:
    for i in range(3):
        product_name = input(f"Enter Product Name {i}")
        product_price = input(f"Enter Product price {i}")
        file.write(f"{product_name} | {product_price}\n")


# TODO: Read and print products.txt with clean formatting
with open("products.txt", "r") as file:
    lines = file.readlines()
    for line in lines:
        product, price = tuple(line.split("|"))
        print(f"Product {product.strip()} has price ₹{price.strip()}")
````

## Task 7: Mini Project - Export Discounted Prices
Given:
```python
prices = {
    "Mouse": 500,
    "Keyboard": 1500,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}
```

1. Ask the user for a discount percentage.
2. Write a report to `discount_report.txt` with each line in this format:
   `Product, Original Price, Discounted Price`
3. Reopen and print the report in the terminal.
4. Also print:
   - Total items
   - Average discounted price

````python
# Task 7 solution
prices = {
    "Mouse": 500,
    "Keyboard": 1500,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}

discount_percent = float(input("Enter discount percentage: "))

with open("discount_report.txt", "w") as file:
    for product, original in prices.items():
        file.write(f"{product}, {original}, {original*(100-discount_percent)/100}\n")

# TODO: Write discount_report.txt with product, original, discounted prices
# TODO: Reopen and print discount_report.txt
with open("discount_report.txt", "r") as file:
    data = file.readlines()
    prices = [float(line.split(",")[1].strip()) for line in data]
    discounted_prices = [float(line.split(",")[2].strip()) for line in data]
    total = sum(prices)
    avg = total/len(prices)
    discounted_avg = sum(discounted_prices)/len(prices)
    
    print(f"Total {total}, avergare {avg}, discounted average: {discounted_avg}")
    
# TODO: Print total items and average discounted price
````
