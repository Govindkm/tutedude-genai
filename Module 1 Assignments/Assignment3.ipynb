## Task 1: Basic Function - Apply Discount

Test it with:

## Task 2: Recursive Function - Factorial
Create a recursive function `factorial(n)` that:
1. Returns `1` when `n` is `0` or `1`.
2. Returns a clear message when `n` is negative.
3. Calls itself for positive values greater than `1`.

Test it with:
- `factorial(5)`
- `factorial(0)`
- `factorial(-3)`

````python
# Task 2 solution
def factorial(n):
    # TODO: Add the base cases and recursive case
    if n==1 or n==0:
        return 1;
    if n<0:
        return ValueError("Invalid input provided")
    return n*factorial(n-1)

# TODO: Test the function with all three examples
print(factorial(5))
print(factorial(0))
print(factorial(-3))
````

## Task 6: Combined Utility Function
Write a function `process_prices(prices, discount_percent=10)` that:
1. Receives a list of prices and a discount percentage.
2. Uses `map()` and `lambda` to apply the discount to every price.
3. Uses `filter()` and `lambda` to keep only discounted prices above `300`.
4. Returns the filtered list.

Test it with:
```python
prices = [100, 500, 250, 1000, 50, 450]
```

````python
# Task 6 solution
def process_prices(prices, discount_percent=10):
    # TODO: Use map() and lambda to apply the discount
    discounted = list(map(lambda price: price * (100-discount_percent)/100, prices))
    # TODO: Use filter() and lambda to keep values above 300
    filtered = list(filter(lambda price: price > 300, discounted))
    # TODO: Return the filtered list
    return filtered

prices = [100, 500, 250, 1000, 50, 450]
# TODO: Call the function and print the result
print(f"Original prices: {prices}\n Discounted and filtered prices: {process_prices(prices)}")
````

## Task 7: Mini Project - Menu Using Functions
Create a menu-driven price utility using functions.

Create these functions:
1. `add_price(prices, price)` - adds a price to the list.
2. `get_average_price(prices)` - returns the average price.
3. `get_max_price(prices)` - returns the highest price.
4. `show_menu()` - displays the available options.

The menu should support:
- `1` - Add a price
- `2` - Show the average price
- `3` - Show the highest price
- `q` - Quit

Use a loop and function calls to keep the menu running until the user chooses `q`.

````python
# Task 7 solution
prices_list = []

def add_price(prices: list, price):
    # TODO: Add the price and return the updated list
    prices.append(price)

def get_average_price(prices):
    # TODO: Return the average price
    total = sum(prices)
    length = len(prices)
    return total/length

def get_max_price(prices):
    # TODO: Return the highest price
    return max(prices)

def show_menu():
    # TODO: Print the menu options
    print(f"""
Here are your menu options:
    1. Add a price
    2. Show the average price
    3. Show the highest price
    q. Quit
          """)

# TODO: Build the menu loop using the functions above
# Use input(), if/elif/else, continue, and break where appropriate.
while(True):
    show_menu()
    user_input = input("Please select appropriate option from the menu")
    if user_input.lower() == 'q':
        break
    elif user_input == '1':
        try:
            user_input = float(input("Enter price value to add"))
            user_input = int(user_input) if user_input.is_integer() else user_input
            add_price(prices_list, user_input)
        except:
            print("Please give a numeric value")
            continue
    elif user_input == '2':
        print(f"Average price for {prices_list} is {get_average_price(prices_list)}")
    elif user_input == '3':
        print(f"Highest amongst {prices_list} is {get_max_price(prices_list)}")
    else:
        print("Select proper input!!!")
        continue
````
