## Task 1: Discount Rules (if/elif/else)
1. Write a program that reads an integer `order_amount` from the user using `input()`.
2. Apply the following discount rules and print the final amount:
   - `order_amount >= 2000` -> 15% discount
   - `1500 <= order_amount < 2000` -> 10% discount
   - `1000 <= order_amount < 1500` -> 7% discount
   - `order_amount < 1000` -> 0% discount
3. Ensure input is converted to a number and handle non-numeric input by displaying an error message and exiting the program.

**Extra (optional):** Add tax (5%) after the discount is applied and print the subtotal (after discount), tax, and final total.


```python
# Task 1 solution
# TODO: Read order amount from user
try:
    order_amount = float(input("Enter order amount"))
# TODO: Apply discount with if/elif/else
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0.0

    subtotal = order_amount * (1 - discount_rate)
# TODO (optional): Add tax after discount
    tax = subtotal * 0.05
    final_total = subtotal + tax
# TODO: Print discount and final amount
    print(f"Subtotal after discount: {subtotal}")
    print(f"Tax (5%): {tax}")
    print(f"Final total: {final_total}")
# TODO: Handle invalid (non-numeric) input
except ValueError:
    print("Please enter a numeric value")
```

## Task 2: Process Multiple Orders (for loop)
1. Given a list of order amounts `[1200, 2500, 880, 17500, 3000]`, use an explicit `for` loop (no list comprehensions) to:
   - Apply the discount rules from Task 1 for each amount
   - Print a summary table with clear columns: `order_amount -> discount% -> final_amount`
2. Compute and print total revenue after discounts.

**Extra (optional):** Print the number of orders that received a discount (`discount > 0`).


```python
# Task 2 solution
orders = [1200, 2500, 880, 17500, 3000]

total_revenue = 0
discounted_count = 0

# TODO: Loop through orders
print(f"{'order_amount':>15} -> {'discount%':>10} -> {'final_amount':>15}")
for order_amount in orders:
# TODO: Apply discount rules
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0.0

    final_amount = order_amount * (1 - discount_rate)
    total_revenue = total_revenue + final_amount

# TODO (optional): Count discounted orders
    if discount_rate > 0:
        discounted_count = discounted_count + 1

# TODO: Print summary rows
    print(f"{order_amount:>15} -> {discount_rate * 100:>9.0f}% -> {final_amount:>15.2f}")

# TODO: Compute total revenue after discounts
print(f"\nTotal revenue after selling: {total_revenue}")
print(f"Total number of discounted orders: {discounted_count}")
```

## Task 4: Data Validation and Skip Bad Data (for + continue + break)
Given the list values:
`[200, 150, 0, 400, 50, -1, 300]`

1. Iterate through the list and:
   - If a value is `-1`, treat it as corrupted data and `break` the loop
   - If a value is `0`, treat it as a no-sale day: `continue` and skip adding it to the total
   - For valid values, add to `total_sales` and print the running total
2. Print the final total after the loop completes (or stops due to corrupted data).


```python
# Task 4 solution
values = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0
for item in values:
# TODO: break on corrupted data (-1)
    if item == -1:
        print("Corrupted data (-1) encountered, stopping.")
        break
# TODO: continue on no-sale day (0)
    if item == 0:
        print("No sale day, skipping.")
        continue
# TODO: Add valid values and print running total
    total_sales = total_sales + item
    print(f"Running total: {total_sales}")

# TODO: Print final total
print(f"Final total: {total_sales}")
```
