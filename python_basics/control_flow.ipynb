## Control Flow: if, elif, else

**Analogy:** Think of traffic lights. If green, go. If yellow, slow down. Else (red), stop.

**What it is:** Control flow lets your program choose different actions based on conditions.

```python
# Minimal working example
temperature = 31

if temperature >= 35:
    print("It is very hot")
elif temperature >= 25:
    print("It is warm")
else:
    print("It is cool")
```

- If score is 90 or more, print `Excellent`
- If score is between 60 and 89, print `Good job`
- Else print `Keep practicing`

**Hint:** Start by creating `score = 75`.

```python
# YOUR CODE HERE
score = 75

# Write if-elif-else using the rules above
if score >= 90:
    print("Excellent")
elif score >= 60:
    print("Good Job")
else:
    print("Keep Practicing")
```

### Watch out for

- Using `=` instead of `==` in conditions
  - Wrong: `if score = 90:`
  - Right: `if score == 90:`
- Wrong indentation after `if`, `elif`, `else`
  - Python uses indentation to define blocks

### Quick Revision

| What | How |
|------|-----|
| `if` | Runs when first condition is true |
| `elif` | Checks next condition if previous one failed |
| `else` | Runs when none of the above conditions match |
