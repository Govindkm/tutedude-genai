## File I/O and Modules

**Analogy:** A file is like a notebook on your desk. You can open it, read pages, write new pages, and close it safely.

**What it is:** File I/O lets Python save and load data from files so your data stays even after the program ends.

```python
# Minimal working example - run this first!
with open('notes.txt', 'w') as f:
    f.write('Python File I/O is useful.\n')
    f.write('Second line.\n')

with open('notes.txt', 'r') as f:
    content = f.read()

print(content)
```

## Assignment: File I/O Operations (Mini Project)

### Scenario
You are building a tiny **Study Tracker** for a student. The tracker stores tasks in a text file and generates a daily summary report.

### Learning Goals
- Use `w`, `a`, and `r` modes correctly
- Process file content line-by-line
- Filter and transform file data
- Write computed results to a new output file

### Rules
- Solve using File I/O (no external libraries)
- Use functions for each task
- Keep code readable and test with the self-check cell

```python
# Starter dataset for your assignment
# Format per line: task_name|category|minutes|status
# status is either done or pending

sample_lines = [
    "Read Python chapter|study|45|done",
    "Practice loops|study|30|done",
    "Watch ML video|learning|40|pending",
    "Solve 5 exercises|practice|50|done",
    "Revise notes|study|20|pending",
]

with open("study_log.txt", "w") as f:
    for line in sample_lines:
        f.write(line + "\n")

print("Created study_log.txt with sample data")
```

### Tasks

1. **Append new tasks**  
Create a function `add_task(task, category, minutes, status)` that appends one line to `study_log.txt` in the same `|` format.

2. **Load and parse data**  
Create a function `load_tasks()` that reads the file and returns a list of dictionaries like:
`{"task": ..., "category": ..., "minutes": int, "status": ...}`

3. **Filter completed tasks**  
Create `completed_tasks(tasks)` that returns only tasks with status `done`.

4. **Compute summary**  
Create `build_summary(tasks)` that returns a dictionary with:
- `total_tasks`
- `done_tasks`
- `pending_tasks`
- `total_minutes_done`
- `study_minutes_done` (only category `study` and status `done`)

5. **Write report file**  
Create `write_report(summary)` that writes `daily_report.txt` in a clean format.

6. **Challenge (optional)**  
Create `high_focus_tasks(tasks, min_minutes=40)` that returns task names with `minutes >= min_minutes` and `status == done`.

```python
# TODO: Implement all functions below

def add_task(task, category, minutes, status):
    with open("study_log.txt", "a") as f:
        f.write(f"{task}|{category}|{minutes}|{status}\n")

def load_tasks():
    with open("study_log.txt", "r") as f:
        data = f.readlines()
        print(f"Loaded {len(data)} tasks from study_log.txt")
    tasks = []
    for line in data:
        item = line.strip().split("|")
        if len(item) == 4:
            tasks.append({
                "task": item[0],
                "category": item[1],
                "minutes": int(item[2]),
                "status": item[3],
            })
    return tasks


def completed_tasks(tasks):
    completed = [task for task in tasks if task["status"] == "done"]
    return completed


def build_summary(tasks):
    result = {}
    result["total_tasks"] = len(tasks)
    result["done_tasks"] = [task for task in tasks if task["status"] == "done"]
    result["pending_tasks"] = [task for task in tasks if task["status"] == "pending"]

    result["total_minutes_done"] = sum(task["minutes"] for task in result["done_tasks"])
    result["study_minutes_done"] = sum(
        task["minutes"] for task in result["done_tasks"] if task["category"] == "study"
    )
    return result


def write_report(summary):
    with open("daily_report.txt", "w") as f:
        f.write(f"Total Tasks: {summary['total_tasks']}\n")
        f.write(f"Done Tasks: {len(summary['done_tasks'])}\n")
        f.write(f"Pending Tasks: {len(summary['pending_tasks'])}\n")
        f.write(f"Total Minutes Done: {summary['total_minutes_done']}\n")
        f.write(f"Study Minutes Done: {summary['study_minutes_done']}\n")
    print("Report written to daily_report.txt")


def high_focus_tasks(tasks, min_minutes=40):
    # Return only task names that are both done and above threshold.
    focus_tasks = [
        task["task"]
        for task in tasks
        if task["minutes"] >= min_minutes and task["status"] == "done"
    ]
    return focus_tasks


# --------------------
# Self-check (do not edit)
# --------------------
add_task("Mock interview", "practice", 60, "done")

tasks = load_tasks()
print(tasks)
assert isinstance(tasks, list), "load_tasks should return a list"
assert len(tasks) >= 6, "After add_task, there should be at least 6 tasks"
assert all(isinstance(t.get("minutes"), int) for t in tasks), "minutes should be int"

done = completed_tasks(tasks)
assert all(t["status"] == "done" for t in done), "completed_tasks should only keep done tasks"

summary = build_summary(tasks)
required_keys = {
    "total_tasks",
    "done_tasks",
    "pending_tasks",
    "total_minutes_done",
    "study_minutes_done",
}
assert required_keys.issubset(summary.keys()), "Missing keys in summary"

write_report(summary)
with open("daily_report.txt", "r") as f:
    report_text = f.read()
assert "Total Tasks" in report_text, "Report should contain Total Tasks"

focus = high_focus_tasks(tasks, 40)
assert isinstance(focus, list), "high_focus_tasks should return a list"
assert all(isinstance(name, str) for name in focus), "high_focus_tasks should return task names"

print("All checks passed. Great job!")
```

## Module Exercise Ladder

### 🟢 Beginner
Build a function `circle_stats(radius)` using `math` that returns:
- area (rounded to 2 decimals)
- circumference (rounded to 2 decimals)

### 🟡 Intermediate
Build a function `practice_plan(seed=42)` using `random` that:
- sets a seed for repeatable output
- returns a list of 5 random integers from 1 to 10

### 🔴 Challenge
Build a function `days_left_in_year()` using `datetime` that returns how many days are left from today until Dec 31 of the current year.

Try to complete the next code cell before looking at any help.

```python
import math
import random
import datetime as dt

# YOUR CODE HERE

def circle_stats(radius):
    # Return (area, circumference), both rounded to 2 decimals.
    area = round(math.pi * radius ** 2, 2)
    circumference = round(2 * math.pi * radius, 2)
    return (area, circumference)


def practice_plan(seed=42):
    # Use a fixed seed for repeatable random output.
    random.seed(seed)
    return [random.randint(1, 10) for _ in range(5)]


def days_left_in_year():
    # Compute remaining days from today to Dec 31.
    today = dt.date.today()
    last_day_of_year = dt.date(today.year, 12, 31)
    left = (last_day_of_year - today).days
    return left


# Quick self-check (run after implementing)
# Expected shape checks
a, c = circle_stats(7)
assert isinstance(a, float) and isinstance(c, float)
#
plan = practice_plan(42)
assert isinstance(plan, list) and len(plan) == 5
assert all(1 <= x <= 10 for x in plan)
#
left = days_left_in_year()
assert isinstance(left, int) and left >= 0
#
print("Module exercise checks passed!")
```

## Module Exercise Level-Up

### Scenario
You are building a tiny daily planner helper for study sessions.

### 🟢 Beginner
Create `study_blocks(total_minutes, block_size=25)` using `math`.
- Return number of blocks needed (use `math.ceil`).

### 🟡 Intermediate
Create `pick_focus_topic(topics, seed=7)` using `random`.
- Set seed for deterministic behavior.
- Return one chosen topic from the list.

### 🔴 Challenge
Create `session_stamp()` using `datetime`.
- Return a string in this exact format: `YYYY-MM-DD HH:MM`.
- Use current local date and time.

```python
import math
import random
import datetime as dt

# YOUR CODE HERE

def study_blocks(total_minutes, block_size=25):
    # Return number of blocks needed.
    totalBlocks = total_minutes / block_size
    return math.ceil(totalBlocks)


def pick_focus_topic(topics, seed=7):
    # Return one deterministic random topic.
    random.seed(seed)
    randomPos = random.randint(0, len(topics)-1)
    return topics[randomPos]


def session_stamp():
    # Return current timestamp in format YYYY-MM-DD HH:MM
    return dt.datetime.today().isoformat(sep=" ", timespec="minutes")


# Self-check
assert study_blocks(60, 25) == 3
assert pick_focus_topic(["Python", "ML", "SQL"], seed=7) in ["Python", "ML", "SQL"]
stamp = session_stamp()
print(stamp)
assert isinstance(stamp, str)
assert len(stamp) == 16 and stamp[4] == "-" and stamp[7] == "-" and stamp[10] == " " and stamp[13] == ":"

print("Level-up checks passed!")
```
