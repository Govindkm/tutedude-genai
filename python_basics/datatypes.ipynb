# Python Data Types

## Primitive (Basic) Data Types

### 1. Integers (`int`)
- Whole numbers, positive or negative, without decimals
- Example: `x = 10`, `y = -3`
- Python supports arbitrarily large integers

### 2. Floating-Point Numbers (`float`)
- Numbers with decimal points
- Example: `x = 3.14`, `y = -0.001`
- Can also use scientific notation: `x = 1.5e2` (150.0)

### 3. Booleans (`bool`)
- Represents `True` or `False`
- Subclass of `int` (`True == 1`, `False == 0`)
- Example: `is_valid = True`

### 4. Strings (`str`)
- Sequence of characters enclosed in quotes
- Example: `name = "Alice"`, `msg = 'Hello'`
- Immutable — once created, individual characters cannot be changed

### 5. NoneType (`None`)
- Represents the absence of a value
- Example: `result = None`
- There is only one instance of `None`

### 6. Complex Numbers (`complex`)
- Numbers with a real and imaginary part
- Example: `z = 3 + 4j`

---

## Reference (Non-Primitive / Compound) Data Types

### 1. Lists (`list`)
- Ordered, mutable collection of items
- Example: `nums = [1, 2, 3]`
- Can hold mixed types: `mixed = [1, "hello", True]`

### 2. Tuples (`tuple`)
- Ordered, **immutable** collection of items
- Example: `point = (3, 4)`
- Faster than lists due to immutability

### 3. Dictionaries (`dict`)
- Unordered collection of key-value pairs
- Example: `person = {"name": "Alice", "age": 25}`
- Keys must be immutable (strings, numbers, tuples)

### 4. Sets (`set`)
- Unordered collection of **unique** items
- Example: `unique_nums = {1, 2, 3}`
- Supports set operations like union, intersection

### 5. Frozen Sets (`frozenset`)
- Immutable version of a set
- Example: `fs = frozenset([1, 2, 3])`

### 6. Byte Types (`bytes`, `bytearray`, `memoryview`)
- `bytes`: Immutable sequence of bytes — `b"hello"`
- `bytearray`: Mutable sequence of bytes — `bytearray(5)`
- `memoryview`: Memory view of a bytes-like object

## Key Difference: Primitive vs Reference Types

| Feature | Primitive Types | Reference Types |
|---|---|---|
| **Storage** | Store the actual value | Store a reference (pointer) to the object |
| **Mutability** | Immutable (`int`, `float`, `str`, `bool`, `tuple`) | Often mutable (`list`, `dict`, `set`) |
| **Assignment** | Creates a copy of the value | Creates a new reference to the same object |
| **Examples** | `int`, `float`, `str`, `bool`, `None` | `list`, `dict`, `set`, `tuple`, custom objects |

> **Note:** Technically, *everything* in Python is an object, so even "primitive" types are objects with references. The distinction is more about **mutability** and **behavior** than strict memory semantics as in languages like Java or C.
