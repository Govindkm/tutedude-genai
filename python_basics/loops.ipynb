## Loops in Python (for and while)

**Analogy:** A loop is like doing daily push-ups. You repeat the same action for each count until you finish.

**What it is:** Loops help you run the same block of code multiple times without rewriting it.

```python
# Live demo: for loop and while loop
print("for loop demo")
for i in range(1, 6):
    print("Count:", i)

print("\nwhile loop demo")
n = 1
while n <= 3:
    print("Step:", n)
    n += 1
```

### Your Turn

**Problem:** You are tracking 5 practice scores.
1. Use a `for` loop to print each score and whether it is `Pass` (>= 60) or `Fail`.
2. Use a `while` loop to print numbers from 3 down to 1, then print `Go!`.

**Hint:** Start with `scores = [95, 72, 58, 89, 40]`.

```python
# YOUR CODE HERE
scores = [95, 72, 58, 89, 40]

# Part 1: for loop with Pass/Fail
# for score in scores:
#     ...

for score in scores:
    if score >= 60:
        print(f"Scored : {score}, Result: Passed")
    else:
        print(f"Scored: {score}, Result: Try again")

# Part 2: while loop countdown 3, 2, 1, Go!
# n = 3
# while n >= 1:
#     ...
i = 3
while(i>=0):
    if(i>0):
        print(i)
    else:
        print('Go!')
    i = i - 1
```

### Watch out for

- Forgetting to update the variable in `while` loops can create an infinite loop.
- Off-by-one errors with `range()` are common.
  - `range(1, 6)` gives 1 to 5, not 1 to 6.
- Keep indentation consistent inside the loop block.

### Quick Revision

| What | How |
|------|-----|
| `for` loop | Repeats over items in a sequence like list/range |
| `while` loop | Repeats while a condition stays true |
| Loop control | Update condition variables to avoid infinite loops |

## Nested Loops

**Analogy:** Think of a classroom with rows and columns of seats. For each row, you visit every seat in that row.

**What it is:** A nested loop is a loop inside another loop. The inner loop runs fully for each outer loop step.

```python
# Live demo: nested loops
rows = [1, 2, 3]
cols = [1, 2, 3]

for r in rows:
    for c in cols:
        print(f"Cell({r},{c})")
```

### Your Turn: Nested Loops

**Problem:** Print a 1 to 3 multiplication table using nested loops.

Expected style:
- 1 x 1 = 1
- 1 x 2 = 2
- ... up to 3 x 3 = 9

**Hint:** Outer loop for first number, inner loop for second number.

```python
# YOUR CODE HERE
for i in range(1, 4):
    for j in range(1, 4):
        # print in format: i x j = result
        print(f"{i} x {j} = {i*j}")
    print()
```

### Watch out for

- Reusing the same variable name in both loops can confuse output.
- Indentation defines which statements belong to inner vs outer loop.
- Nested loops increase total iterations quickly, so keep ranges small while learning.

### Quick Revision

| What | How |
|------|-----|
| Outer loop | Controls the larger repeated group |
| Inner loop | Runs fully for each outer loop iteration |
| Typical use | Tables, grid traversal, pair combinations |

## Nested Loops with continue and break

### Task A: Use continue
Print pairs `(i, j)` for `i` and `j` from 1 to 4, but skip pairs where `i == j`.

### Task B: Use break
For each `i` from 1 to 4, loop `j` from 1 to 4 and stop inner loop when `j == 3`.
Print values before break in this format: `(i:X,j:Y)`.

```python
# Task A
for i in range(1, 5):
    for j in range(1, 5):
        if(i==j):
            continue
        print(f"(i:{i},j:{j})")
```

```python
# Task B
for i in range(1, 5):
    for j in range(1, 5):
        if(j==3):
            break
        print(f"(i:{i},j:{j})")
```
