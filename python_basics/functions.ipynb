## Functions in Python

**Analogy:** A function is like a reusable recipe. You write the steps once, then use it whenever needed.

**What it is:** A function is a named block of code that performs a task and can be called multiple times.

```python
# Live demo: defining and calling functions
def greet(name):
    return f"Hello, {name}!"

def add(a, b):
    return a + b

print(greet("Govin"))
print("2 + 3 =", add(2, 3))
```

**Problem:** Build two small functions for a marks app.
1. `is_pass(score)` should return `True` if score is 60 or above, else `False`.
2. `grade(score)` should return:
   - `A` for 90+
   - `B` for 75 to 89
   - `C` for 60 to 74

**Hint:** Use `return` in each branch.

```python
# YOUR CODE HERE
def is_pass(score):
    return score >= 60

def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 75:
        return 'B'
    elif score >= 60:
        return 'C'
    else:
        return 'F'

# quick checks
print(is_pass(72))   # expected: True
print(is_pass(40))   # expected: False
print(grade(95))     # expected: A
print(grade(78))     # expected: B
print(grade(63))     # expected: C
print(grade(50))     # expected: F
```

## Functions: Default Parameters and Multiple Returns

**Analogy:** A restaurant combo has default sides, but you can customize when needed. Also, one order can return multiple items together.

**What it is:**
- Default parameters let a function use fallback values when arguments are not provided.
- A function can return multiple values as a tuple.

```python
# Live demo: default parameters and multiple return values

def welcome(name, city="Chennai"):
    return f"Hi {name}, welcome from {city}!"

def score_stats(maths, science, english):
    total = maths + science + english
    average = total / 3
    return total, round(average, 2)

print(welcome("Asha"))
print(welcome("Ravi", "Bengaluru"))

total, avg = score_stats(80, 75, 90)
print("Total:", total)
print("Average:", avg)
```

### Your Turn

**Problem:** Build two functions for a shopping app.

1. `final_price(price, discount=0)`
- Apply discount percentage and return final price.
- Example: `final_price(1000, 10)` should return `900.0`.

2. `min_max(numbers)`
- Return both minimum and maximum from a list as two values.
- Example: `min_max([5, 1, 9, 3])` should return `1, 9`.

**Hint:**
- Formula: `price - (price * discount / 100)`
- Use `return min_value, max_value`

```python
# YOUR CODE HERE

def final_price(price, discount=0):
    # return the discounted final amount
    return price*(100-discount)/100

def min_max(numbers):
    # return minimum and maximum as two values
    max_value = max(numbers)
    min_value = min(numbers)
    return min_value, max_value

# quick checks
print(final_price(1000))         # expected: 1000.0
print(final_price(1000, 10))     # expected: 900.0

small, big = min_max([5, 1, 9, 3])
print(small, big)                # expected: 1 9
```

### Watch out for

- Default parameter values should usually be simple immutable values (`0`, `""`, `None`).
- When returning multiple values, Python packs them as a tuple.
- Unpack in same order: `a, b = function_call(...)`.

### Quick Revision

| What | How |
|------|-----|
| Default parameter | Used when caller does not provide that argument |
| Multiple return values | `return x, y` and receive via `a, b = ...` |
| Benefit | Cleaner reusable functions with flexible inputs |

## Functions: Keyword Arguments and *args

**Analogy:** Ordering food by saying exactly which item is which is like keyword arguments. Saying "bring all these toppings" is like `*args`.

**What it is:**
- Keyword arguments let you pass values using parameter names.
- `*args` allows a function to accept any number of positional arguments.

```python
# Live demo: keyword args and *args

def student_card(name, age, city):
    return f"{name} | {age} | {city}"

def total_marks(*marks):
    return sum(marks)

print(student_card(name="Asha", age=14, city="Chennai"))
print(student_card(city="Pune", name="Ravi", age=15))

print(total_marks(80, 75, 90))
print(total_marks(60, 70, 85, 95))
```

### Your Turn

**Problem:** Build two functions for an app profile and score summary.

1. `profile(name, city="Chennai", age=18)`
- Return string format: `name - age - city`
- Call once with all keyword arguments in different order.

2. `best_score(*scores)`
- Return highest score from given arguments.
- Example: `best_score(65, 88, 72, 91)` should return `91`.

**Hint:**
- Use `max(scores)` for `*scores`.
- Try one call with default city and age.

```python
# YOUR CODE HERE

def profile(name, city="Chennai", age=18):
    # return in format: name - age - city
    return f"{name} - {city} - {age}"

def best_score(*scores):
    # return highest score
    return max(scores)

# quick checks
print(profile("Mina"))                                   # expected: Mina - 18 - Chennai
print(profile(city="Pune", name="Karan", age=20))      # expected: Karan - 20 - Pune

print(best_score(65, 88, 72, 91))                         # expected: 91
```

### Watch out for

- Keyword argument names must match parameter names exactly.
- In function definitions, normal parameters come before `*args`.
- `*args` is a tuple, so you can use tuple operations like `sum`, `max`, and iteration.

### Quick Revision

| What | How |
|------|-----|
| Keyword arguments | Pass values by name, order does not matter |
| `*args` | Accepts variable number of positional arguments |
| Benefit | Flexible function calls and cleaner reusable APIs |

## Functions: `**kwargs`

**Analogy:** `**kwargs` is like a form where people can fill in different named fields. You do not know all the fields in advance, but you still want to collect them neatly.

**What it is:** `**kwargs` lets a function accept any number of extra keyword arguments and stores them in a dictionary.

```python
# Live demo: collecting keyword arguments

def student_profile(name, **details):
    print("Name:", name)
    print("Other details:", details)

student_profile("Asha", age=14, city="Chennai", grade=9)

def build_message(**kwargs):
    return f"Received {len(kwargs)} extra fields"

print(build_message(course="Python", level="Beginner", batch=3))
```

### Your Turn

**Problem:** Build a function `student_card(name, **details)`.
- Print the student name.
- Print every key and value inside `details`.

Then call it with at least 3 extra keyword arguments, such as `age`, `city`, and `course`.

**Hint:** Use a loop like `for key, value in details.items():`.

```python
# YOUR CODE HERE

def student_card(name, **details):
    print("Student:", name)
    for key, value in details.items():
        print(f"{key}: {value}")

student_card("Govind", age=21, city="Pune", course="Python")
```

### Watch out for

- `**kwargs` becomes a dictionary, so use `.items()` to loop through it.
- The double star `**` is only for keyword arguments, not plain positional values.
- You can combine normal parameters, `*args`, and `**kwargs` in one function when needed.

### Quick Revision
| What | How |
|------|-----|
| `**kwargs` | Collects extra keyword arguments into a dictionary |
| Access | Use `kwargs['key']` or loop with `kwargs.items()` |
| Benefit | Lets functions accept flexible named inputs |
