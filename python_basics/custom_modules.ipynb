## Creating Your Own Module

**Analogy:** A module is like your personal toolbox. You keep useful tools in one place and reuse them in many programs.

**What it is:** A module is a Python file containing reusable functions, variables, or constants that can be imported into other files.

```python
# Minimal working example
# Step 1: Create a custom module file
module_code = '''
PI = 3.14159

def square(n):
    return n * n

def circle_area(r):
    return PI * r * r
'''

with open('my_tools.py', 'w') as f:
    f.write(module_code)

# Step 2: Import and use it
import my_tools

print('Square of 6:', my_tools.square(6))
print('Area of radius 3:', round(my_tools.circle_area(3), 2))
```

## Level 2: from-import, Aliases, and Reload

**Analogy:**
- `import module` is like carrying the whole toolbox.
- `from module import tool` is like taking only one screwdriver out.
- `import module as m` is giving a nickname to your toolbox.
- `reload(module)` is refreshing the toolbox after you changed tools inside it.

**When to use what:**
- Use `import module` for clarity and safer names.
- Use `from module import name` when you need short direct calls.
- Use `as` alias when names are long (like `import datetime as dt`).
- Use `importlib.reload()` in notebooks after editing a module file.

```python
# Live Demo: all three styles + reload

# Step 1: Create a module file
level2_code = '''
def greet(name):
    return f"Hello, {name}!"

def add(a, b):
    return a + b
'''

with open("level2_tools.py", "w") as f:
    f.write(level2_code)

# Step 2: import module style
import level2_tools
print(level2_tools.greet("Govind"))
print(level2_tools.add(10, 5))

# Step 3: from-import style
from level2_tools import greet
print(greet("Learner"))

# Step 4: alias style
import level2_tools as l2
print(l2.add(7, 8))

# Step 5: reload after updating module code
updated_code = '''
def greet(name):
    return f"WELCOME, {name.upper()}!"

def add(a, b):
    return a + b
'''

with open("level2_tools.py", "w") as f:
    f.write(updated_code)

import importlib
importlib.reload(level2_tools)
print(level2_tools.greet("Govind"))
```

### Watch out for (Level 2)

### Quick Revision
| Pattern | Example | Use Case |
|---|---|---|

| Alias import | `import datetime as dt` | Readable short name |
| Reload | `importlib.reload(my_module)` | Pick up edited module code in notebook |

## Advanced Level: Package-Style Imports

**Analogy:** A package is like a labeled drawer with smaller boxes inside it. The drawer is the package folder, and each smaller box is a module file.

**What it is:** A package is a folder of Python modules, usually with an `__init__.py` file, so you can organize related code and import it neatly.

### Your Turn (Package Style)

Create a package named `school_tools` with this structure:
- `school_tools/__init__.py`
- `school_tools/grading.py`
- `school_tools/schedule.py`

Add these functions:
- `grading.py`: `grade(score)` returns `A`, `B`, `C`, or `F`

Challenge: export `grade` from `__init__.py` so you can use `school_tools.grade(88)` directly.

```python
# Package-style demo: create a folder package and import from it

package_init = '''
from .math_tools import double, triple
'''

math_tools_code = '''
def double(n):
    return n * 2

def triple(n):
    return n * 3
'''

import os
os.makedirs("study_package", exist_ok=True)

with open("study_package/__init__.py", "w") as f:
    f.write(package_init)

with open("study_package/math_tools.py", "w") as f:
    f.write(math_tools_code)

import study_package
from study_package import double
import study_package.math_tools as mt

print(study_package.double(4))
print(double(5))
print(mt.triple(6))
```

### Watch out for (Package Style)
- A folder becomes a package when it contains `__init__.py`.
- Use relative imports inside packages when linking one module to another.
- Package names should stay short, clear, and lowercase.

### Quick Revision
| Pattern | Example | Use Case |
|---|---|---|
| Package import | `import study_package` | Access exported tools through the package |
| Module import | `import study_package.math_tools as mt` | Call module functions with a short alias |
| From package import | `from study_package import double` | Direct access to exported names |
| Export in __init__ | `from .math_tools import double` | Make package-level shortcuts |
