# Operators in Python

```python
# ============================================
# Operators in Python
# ============================================

# 1. Arithmetic Operators
print("=" * 50)
print("1. ARITHMETIC OPERATORS")
print("=" * 50)
a, b = 10, 3
print(f"a = {a}, b = {b}")
print(f"Addition:       a + b = {a + b}")
print(f"Subtraction:    a - b = {a - b}")
print(f"Multiplication: a * b = {a * b}")
print(f"Division:       a / b = {a / b}")
print(f"Floor Division: a // b = {a // b}")
print(f"Modulus:        a % b = {a % b}")
print(f"Exponent:       a ** b = {a ** b}")

# 2. Comparison (Relational) Operators
print(f"\n{'=' * 50}")
print("2. COMPARISON OPERATORS")
print("=" * 50)
x, y = 5, 10
print(f"x = {x}, y = {y}")
print(f"Equal:            x == y → {x == y}")
print(f"Not Equal:        x != y → {x != y}")
print(f"Greater Than:     x > y  → {x > y}")
print(f"Less Than:        x < y  → {x < y}")
print(f"Greater or Equal: x >= y → {x >= y}")
print(f"Less or Equal:    x <= y → {x <= y}")

# 3. Logical Operators
print(f"\n{'=' * 50}")
print("3. LOGICAL OPERATORS")
print("=" * 50)
p, q = True, False
print(f"p = {p}, q = {q}")
print(f"p and q → {p and q}")
print(f"p or q  → {p or q}")
print(f"not p   → {not p}")

# 4. Assignment Operators
print(f"\n{'=' * 50}")
print("4. ASSIGNMENT OPERATORS")
print("=" * 50)
n = 20
print(f"n = {n}")
n += 5;  print(f"n += 5  → {n}")
n -= 3;  print(f"n -= 3  → {n}")
n *= 2;  print(f"n *= 2  → {n}")
n /= 4;  print(f"n /= 4  → {n}")
n //= 3; print(f"n //= 3 → {n}")
n **= 2; print(f"n **= 2 → {n}")
n = 10; n %= 3; print(f"n=10; n %= 3 → {n}")

# 5. Bitwise Operators
print(f"\n{'=' * 50}")
print("5. BITWISE OPERATORS")
print("=" * 50)
m, k = 12, 10  # binary: 1100, 1010
print(f"m = {m} (binary: {bin(m)}), k = {k} (binary: {bin(k)})")
print(f"AND:         m & k  = {m & k}  (binary: {bin(m & k)})")
print(f"OR:          m | k  = {m | k}  (binary: {bin(m | k)})")
print(f"XOR:         m ^ k  = {m ^ k}   (binary: {bin(m ^ k)})")
print(f"NOT:         ~m     = {~m} (binary: {bin(~m)})")
print(f"Left Shift:  m << 2 = {m << 2} (binary: {bin(m << 2)})")
print(f"Right Shift: m >> 2 = {m >> 2}   (binary: {bin(m >> 2)})")

# 6. Identity Operators
print(f"\n{'=' * 50}")
print("6. IDENTITY OPERATORS")
print("=" * 50)
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1
print(f"list1 = {list1}, list2 = {list2}, list3 = list1")
print(f"list1 is list2  → {list1 is list2}   (same values, different objects)")
print(f"list1 is list3  → {list1 is list3}    (same object)")
print(f"list1 is not list2 → {list1 is not list2}")

# 7. Membership Operators
print(f"\n{'=' * 50}")
print("7. MEMBERSHIP OPERATORS")
print("=" * 50)
fruits = ["apple", "banana", "cherry"]
print(f"fruits = {fruits}")
print(f"'banana' in fruits     → {'banana' in fruits}")
print(f"'grape' in fruits      → {'grape' in fruits}")
print(f"'grape' not in fruits  → {'grape' not in fruits}")
```
