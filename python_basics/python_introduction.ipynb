2. Indentation in Python

In other programming languages such as C++, Java or JS, a block(group) of code is managed using curly braces({}).

In python we use Indentation to group common code into groups with proper spacing and without indentation compiler won't be able to read the code and gives error.

```python
# Working example (correct indentation)
print("Working example:")
if name == "Govind":
    print(f"Hello, {name}!")
    print(f"Full name is: {Name}")
else:
    print("Name did not match.")

# Error example (incorrect indentation) shown safely using exec
print("\nExample that throws IndentationError:")
bad_code = """
if name == "Govind":
print("This line should be indented")
"""

try:
    exec(bad_code)
except IndentationError as e:
    print("IndentationError caught:", e)
```
